```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
hamaker = np.asarray([79.82E-21, 61.67E-21, 52.62E-21, 39.96E-21, 37.26E-21, 33.54E-21])
radii = np.asarray([1E-9, 5E-9, 10E-9, 50E-9, 100E-9, 400E-9])
radii*=2 #we work with diameters
```

```python
fit = np.polyfit(np.log(radii), np.log(hamaker), 3)
fit_r = np.linspace(1E-9, 900E-9, 100)
plt.scatter(radii*1e9, hamaker*1e20)
plt.plot(fit_r*1E9, 1e20*np.exp(np.polyval(fit, np.log(fit_r))))

plt.ylabel(r"Hamaker constant (J$\ \times 10^{-20}$)")
plt.xlabel(r"Diameter [nm]")

print(np.exp(np.polyval(fit, np.log(30E-9))))

```
