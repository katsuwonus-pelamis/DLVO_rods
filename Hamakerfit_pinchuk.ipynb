```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
hamaker = [79.82E-21, 61.67E-21, 52.62E-21, 39.96E-21, 37.26E-21, 33.54E-21]
radii = [1E-9, 5E-9, 10E-9, 50E-9, 100E-9, 400E-9]
```

```python
fit = np.polyfit(np.log(radii), np.log(hamaker), 3)
fit_r = np.linspace(1E-9, 500E-9, 100)
plt.scatter(radii, hamaker)
plt.plot(fit_r, np.exp(np.polyval(fit, np.log(fit_r))))
print(15E-9, np.exp(np.polyval(fit, np.log(15E-9))))
print(fit)

```
